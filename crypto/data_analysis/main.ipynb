```python
def create_event_features_optimized(brti_data, trades_df, orderbook_df, lookback_ms=100):
    """
    Optimized version using vectorized operations
    lookback_ms: lookback window in milliseconds (100ms = 0.1s)
    """
    # Ensure data is sorted by timestamp
    brti_data = brti_data.sort_values('timestamp_ms').reset_index(drop=True)
    trades_df = trades_df.sort_values('timestamp_ms').reset_index(drop=True)
    orderbook_df = orderbook_df.sort_values('timestamp_ms').reset_index(drop=True)
    
    features_list = []
    
    for idx, row in brti_data.iterrows():
        current_time_ms = row['timestamp_ms']
        lookback_time_ms = current_time_ms - lookback_ms
        
        # Use boolean indexing for faster filtering
        trade_mask = (trades_df['timestamp_ms'] >= lookback_time_ms) & (trades_df['timestamp_ms'] <= current_time_ms)
        orderbook_mask = (orderbook_df['timestamp_ms'] >= lookback_time_ms) & (orderbook_df['timestamp_ms'] <= current_time_ms)
        
        recent_trades = trades_df[trade_mask]
        recent_orderbook = orderbook_df[orderbook_mask]
        
        # Quick feature calculation
        buy_vol = recent_trades['buy_volume'].sum()
        sell_vol = recent_trades['sell_volume'].sum()
        total_vol = buy_vol + sell_vol
        
        features = {
            'timestamp_ms': current_time_ms,
            'brti_price': row['brti_price'],
            
            # Core trade features
            'buy_volume_100ms': buy_vol,
            'sell_volume_100ms': sell_vol,
            'net_flow_100ms': buy_vol - sell_vol,
            'flow_ratio_100ms': buy_vol / (total_vol + 1e-8),
            'trade_count_100ms': len(recent_trades),
            
            # Orderbook features
            'spread_mean_100ms': recent_orderbook['spread'].mean() if len(recent_orderbook) > 0 else np.nan,
            'bid_qty_mean_100ms': recent_orderbook['bid_qty'].mean() if len(recent_orderbook) > 0 else np.nan,
            'ask_qty_mean_100ms': recent_orderbook['ask_qty'].mean() if len(recent_orderbook) > 0 else np.nan,
            'orderbook_updates_100ms': len(recent_orderbook),
        }
        
        features_list.append(features)
    
    return pd.DataFrame(features_list).ffill().fillna(0)
```

### parsing tickers into strike dates and strikes (only for kalshi)

```python
import pandas as pd
from datetime import datetime

def parse_ticker(ticker):
    """
    Parse Kalshi ticker format: KXBTCD-xxMONDAYHH-T<strike>
    Example: KXBTCD-25JUN1714-T104249.99
    Returns: dict with strike_date and strike_price
    """
    try:
        # Split the ticker into parts
        parts = ticker.split('-')
        if len(parts) != 3 or not parts[2].startswith('T'):
            raise ValueError(f"Invalid ticker format: {ticker}")
        
        # Extract date and hour part
        date_hour_part = parts[1]  # e.g., "25JUN1714"
        
        # Extract strike price
        strike_price = float(parts[2][1:])  # Remove 'T' prefix and convert to float
        
        # Parse date components
        day = int(date_hour_part[:2])  # First 2 digits
        month_abbr = date_hour_part[2:5]  # Next 3 characters
        year_hour = date_hour_part[5:]  # Remaining characters
        
        # Extract year and hour
        year = int('20' + year_hour[:2])  # Assuming 2-digit year (17 -> 2017)
        hour = int(year_hour[2:])  # EST hour
        
        # Month mapping
        month_map = {
            'JAN': 1, 'FEB': 2, 'MAR': 3, 'APR': 4, 'MAY': 5, 'JUN': 6,
            'JUL': 7, 'AUG': 8, 'SEP': 9, 'OCT': 10, 'NOV': 11, 'DEC': 12
        }
        
        month = month_map.get(month_abbr)
        if month is None:
            raise ValueError(f"Invalid month abbreviation: {month_abbr}")
        
        # Create datetime object (assuming EST timezone)
        strike_date = datetime(year, month, day, hour)
        
        return {
            'strike_date': strike_date,
            'strike_price': strike_price,
            'ticker': ticker
        }
        
    except Exception as e:
        print(f"Error parsing ticker {ticker}: {e}")
        return None

# Test with your example
ticker = "KXBTCD-25JUN1714-T104249.99"
result = parse_ticker(ticker)
print(result)

# Apply to your DataFrame
df['parsed_ticker'] = df['ticker'].apply(parse_ticker)

# Extract strike_date and strike_price into separate columns
df['strike_date'] = df['parsed_ticker'].apply(lambda x: x['strike_date'] if x else None)
df['strike_price'] = df['parsed_ticker'].apply(lambda x: x['strike_price'] if x else None)

# Drop the temporary column
df = df.drop('parsed_ticker', axis=1)
```

### analysis of kalshi data

```python
most_entries = 0
strike_with_most_entries = None
strike_dates_with_most_entries = None

for strike in df.strike_price.unique():
    print(f"Strike Price: {strike}")
    print(f"Number of entries: {len(df[df['strike_price'] == strike])}")

    strike_df = df[df['strike_price'] == strike]

    for date in strike_df['strike_date'].unique():
        # print info on the strike date, like how many rows there are 
        print(f"\tStrike Date: {date}")
        num_entries = len(strike_df[strike_df['strike_date'] == date])
        print(f"\tNumber of entries: {num_entries}")
        if num_entries > most_entries:
            most_entries = num_entries
            strike_with_most_entries = strike
            strike_dates_with_most_entries = date
    
print(f"Strike with most entries: {strike_with_most_entries} on {strike_dates_with_most_entries} with {most_entries} entries")
```
